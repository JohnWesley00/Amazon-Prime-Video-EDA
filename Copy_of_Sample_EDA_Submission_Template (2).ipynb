## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# Create copies of the original datasets
titles_df2 = titles_df.copy()
credits_df2 = credits_df.copy()

# 1. Remove duplicate records
titles_df2.drop_duplicates(inplace=True)
credits_df2.drop_duplicates(inplace=True)

# 2. Remove extra spaces from text columns
for df in [titles_df2, credits_df2]:
    for column in df.select_dtypes(include='object').columns:
        df[column] = df[column].apply(
            lambda x: x.strip() if isinstance(x, str) else x
        )


# Clean genre names
titles_df2['genres'] = titles_df2['genres'].str.replace(
    r"[\[\]']", "", regex=True
)


# 3. Handle missing values in Titles dataset

# Fill missing descriptions
titles_df2['description'] = titles_df2['description'].fillna(
    'No description available'
)

# Fill missing age certifications
titles_df2['age_certification'] = titles_df2['age_certification'].fillna(
    'Unknown'
)

# Replace missing seasons for movies with 0
titles_df2.loc[
    titles_df2['type'] == 'MOVIE', 'seasons'
] = titles_df2.loc[
    titles_df2['type'] == 'MOVIE', 'seasons'
].fillna(0)

# Convert seasons to nullable integer type
titles_df2['seasons'] = titles_df2['seasons'].astype('Int64')

# 4. Display cleaned dataset information
print("Titles Dataset Shape:", titles_df2.shape)
print("Credits Dataset Shape:", credits_df2.shape)

print("\nRemaining Missing Values in Titles:")
print(titles_df2.isnull().sum())

print("\nRemaining Missing Values in Credits:")
print(credits_df2.isnull().sum())

print("\nRemaining Duplicate Rows:")
print("Titles:", titles_df2.duplicated().sum())
print("Credits:", credits_df2.duplicated().sum())
````

### What all manipulations have you done and insights you found?

## What All Manipulations Have You Done and Insights You Found?

### Data Manipulations Performed

1. **Removed Duplicates:** Removed 3 duplicate rows from the Titles dataset and 56 duplicate rows from the Credits dataset.

2. **Removed Extra Spaces:** Cleaned unnecessary spaces from text columns in both datasets.

3. **Handled Missing Values:** Replaced missing descriptions with "No description available" and missing age certifications with "Unknown".

4. **Handled Seasons:** Replaced missing season values for movies with 0 and converted the column into an integer format.

5. **Preserved Other Missing Values:** Kept missing IMDb, TMDb ratings and character information unchanged to avoid adding incorrect data.
